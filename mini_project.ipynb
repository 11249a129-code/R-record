```r

# ============================================
# Experiment: Crime Data Geospatial Analysis
# ============================================

install.packages(c("ggplot2", "dplyr"))
library(ggplot2)
library(dplyr)

crime_data <- data.frame(
  Location = c("Area A","Area B","Area C","Area A","Area B"),
  Latitude = c(13.0827,13.0674,13.0500,13.0827,13.0674),
  Longitude = c(80.2707,80.2376,80.2824,80.2707,80.2376),
  Crime_Type = c("Theft","Assault","Robbery","Theft","Burglary"),
  Time = c("Night","Evening","Night","Morning","Night"),
  Victim_Age = c(25,30,40,22,35)
)

print(crime_data)

crime_count <- crime_data %>%
  group_by(Crime_Type) %>%
  summarise(Count = n())

ggplot(crime_count, aes(x = Crime_Type, y = Count, fill = Crime_Type)) +
  geom_bar(stat = "identity") +
  ggtitle("Crime Type Distribution")

ggplot(crime_data, aes(x = Victim_Age)) +
  geom_histogram(binwidth = 5, fill = "blue") +
  ggtitle("Victim Age Distribution")

ggplot(crime_data, aes(x = Longitude, y = Latitude, color = Crime_Type)) +
  geom_point(size = 3) +
  ggtitle("Crime Locations Map ")
```
